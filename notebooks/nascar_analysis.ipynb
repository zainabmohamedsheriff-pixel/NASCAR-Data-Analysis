```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
plt.scatter(df["LapsLed"], df["Wins"])

plt.title("Laps Led vs Wins")
plt.xlabel("Laps Led")
plt.ylabel("Wins")

plt.show()
```

```python
top_wins = (
    df.groupby("Driver")["Wins"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

top_wins.plot(kind="bar")

plt.title("Top 10 NASCAR Drivers by Total Wins")
plt.xlabel("Driver")
plt.ylabel("Total Wins")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["PercentLapsInTop15"],
    df["Points"]
)

plt.title("Top-15 Consistency vs Championship Points")
plt.xlabel("Percentage of Laps in Top 15")
plt.ylabel("Championship Points")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["PassDiff"],
    df["DriverRating"]
)

plt.title("Passing Performance vs Driver Rating")
plt.xlabel("Pass Difference")
plt.ylabel("Driver Rating")

plt.show()
```

```python
year_rating = df.groupby("Year")["DriverRating"].mean()

plt.figure(figsize=(10, 6))

plt.plot(year_rating.index, year_rating.values, marker="o")

plt.title("Average Driver Rating by Year")
plt.xlabel("Year")
plt.ylabel("Average Driver Rating")

plt.xticks(year_rating.index, rotation=45)

plt.show()
```
